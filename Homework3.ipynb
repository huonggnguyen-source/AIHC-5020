```python
def find_top_student_in_major(df, major_name):

    filtered_df = df[df['Major'] == major_name]


    if filtered_df.empty:
        print(f"No students found for the major: {major_name}")
        return None


    sorted_df = filtered_df.sort_values(by='Score', ascending=False)


    top_student = sorted_df.iloc[0]


    result = top_student[['Name', 'Score']]


    return result
```

```python
import pandas as pd
import matplotlib.pyplot as plt


def plot_histograms_by_category(df, continuous_col, categorical_col):
    if continuous_col not in df.columns:
        print(f"Error: '{continuous_col}' was not found in the DataFrame.")
        return None

    if categorical_col not in df.columns:
        print(f"Error: '{categorical_col}' was not found in the DataFrame.")
        return None

    continuous_unique = df[continuous_col].nunique()
    categorical_unique = df[categorical_col].nunique()

    if continuous_unique < 10:
        print(
            f"Error: '{continuous_col}' appears to be categorical "
            f"because it has only {continuous_unique} unique values."
        )
        return None

    if categorical_unique >= 10:
        print(
            f"Error: '{categorical_col}' appears to be continuous "
            f"because it has {categorical_unique} unique values."
        )
        return None

    fig, ax = plt.subplots(figsize=(10, 6))

    categories = df[categorical_col].dropna().unique()

    for category in categories:
        subset = df[df[categorical_col] == category]
        ax.hist(
            subset[continuous_col].dropna(),
            bins=20,
            alpha=0.6,
            label=f'{categorical_col} = {category}'
        )

    ax.set_title(
        f'Distribution of {continuous_col} by {categorical_col}'
    )
    ax.set_xlabel(continuous_col)
    ax.set_ylabel('Frequency')

    ax.legend()

    plt.show()
```

```python
plot_histograms_by_category(
    stroke_df,
    'age',
    'stroke'
)
```

```python
def calculate_stroke_prevalence(df, group_by_cols):

    for col in group_by_cols:
        if col not in df.columns:
            print(f"Error: '{col}' was not found in the DataFrame.")
            return None


    for col in group_by_cols:
        unique_count = df[col].nunique()

        if unique_count >= 10:
            print(
                f"Error: '{col}' appears to be continuous/cannot be used "
                f"as a categorical grouping column because it has "
                f"{unique_count} unique values."
            )
            return None


    group_sizes = df.groupby(group_by_cols).size()


    stroke_counts = df.groupby(group_by_cols)['stroke'].sum()


    prevalence = stroke_counts / group_sizes


    result = prevalence.to_frame(name='stroke_prevalence')


    result = result.reset_index()

    return result
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler


def preprocess_for_analysis(
    df,
    cols_to_impute,
    cols_to_scale,
    cols_to_encode,
    cols_to_drop
):
    processed_df = df.copy()


    processed_df = processed_df.drop(columns=cols_to_drop)


    for col in cols_to_impute:
        processed_df[col] = processed_df[col].fillna(
            processed_df[col].mean()
        )


    scaler = StandardScaler()

    scaled_values = scaler.fit_transform(processed_df[cols_to_scale])


    scaled_columns = [
        f'{col}_scaled' for col in cols_to_scale
    ]


    scaled_df = pd.DataFrame(
        scaled_values,
        columns=scaled_columns,
        index=processed_df.index
    )


    processed_df = processed_df.drop(columns=cols_to_scale)


    processed_df = pd.concat(
        [processed_df, scaled_df],
        axis=1
    )
    encoded_df = pd.get_dummies(
        processed_df[cols_to_encode],
        drop_first=True
    )

    processed_df = processed_df.drop(columns=cols_to_encode)


    processed_df = pd.concat(
        [processed_df, encoded_df],
        axis=1
    )


    return processed_df
```

```python
impute_list = ['bmi']

scale_list = [
    'age',
    'avg_glucose_level',
    'bmi'
]

encode_list = [
    'gender',
    'ever_married',
    'work_type',
    'Residence_type',
    'smoking_status'
]

drop_list = [
    'id',
    'location_code'
]
```

```python
processed_df = preprocess_for_analysis(
    stroke_df,
    impute_list,
    scale_list,
    encode_list,
    drop_list
)
```
