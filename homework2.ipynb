```python
def to_falsy(value):
    # Check for primitive types
    if isinstance(value, bool):
        return False
    elif isinstance(value, int):
        return 0
    elif isinstance(value, float):
        return 0.0
    elif isinstance(value, str):
        return ""
    elif isinstance(value, complex):
        return 0j
    elif value is None:
        return None
    else:
        print("non primitive type input")
        return None
```

```python
def sort_and_separate_integers(data):
    # Check if input is a list
    if not isinstance(data, list):
        print("not a list")
        return None

    even = []
    odd = []

    # Process each element
    for i in range(len(data)):
        if not isinstance(data[i], int):
            print(f"non-integer entry found at index {i}: {data[i]}")
            continue

        if data[i] % 2 == 0:
            even.append(data[i])
        else:
            odd.append(data[i])

    # Sort the lists
    even.sort()
    odd.sort()

    # Return a tuple
    return (even, odd)
```

```python
def analyze_list_sets(l0, l1):
    # Check that both inputs are lists
    if not isinstance(l0, list):
        print(f"invalid input: {l0}")
        return None

    if not isinstance(l1, list):
        print(f"invalid input: {l1}")
        return None

    # Primitive types
    primitive_types = (int, float, str, bool, complex, type(None))

    valid_l0 = []
    valid_l1 = []

    # Process l0
    for i in range(len(l0)):
        if isinstance(l0[i], primitive_types):
            valid_l0.append(l0[i])
        else:
            print(f"invalid entry at index {i} in l0: {l0[i]}")

    # Process l1
    for i in range(len(l1)):
        if isinstance(l1[i], primitive_types):
            valid_l1.append(l1[i])
        else:
            print(f"invalid entry at index {i} in l1: {l1[i]}")

    # Convert to sets
    set0 = set(valid_l0)
    set1 = set(valid_l1)

    # Perform set operations
    intersection = sorted(set0.intersection(set1), key=str)
    symmetric_difference = sorted(set0.symmetric_difference(set1), key=str)

    return (intersection, symmetric_difference)
```

```python
import numpy as np

def summarize_numpy_array(arr):
    # Validate input
    if (not isinstance(arr, np.ndarray) or
        arr.ndim != 1 or
        not np.issubdtype(arr.dtype, np.floating)):
        print("invalid input")
        return None

    # Check for empty array
    if arr.size == 0:
        return {
            'min': None,
            'mean': None,
            'max': None,
            'std_dev': None,
            'p25': None,
            'p50': None,
            'p75': None,
            'iqr': None
        }

    # Calculate statistics
    p25 = np.percentile(arr, 25)
    p50 = np.percentile(arr, 50)
    p75 = np.percentile(arr, 75)

    return {
        'min': np.min(arr),
        'mean': np.mean(arr),
        'max': np.max(arr),
        'std_dev': np.std(arr),
        'p25': p25,
        'p50': p50,
        'p75': p75,
        'iqr': p75 - p25
    }
```

```python
import numpy as np
import matplotlib.pyplot as plt

def create_scatter_plot(x_arr, y_arr):
    # Validate inputs
    if (not isinstance(x_arr, np.ndarray) or
        not isinstance(y_arr, np.ndarray) or
        x_arr.ndim != 1 or
        y_arr.ndim != 1 or
        not np.issubdtype(x_arr.dtype, np.floating) or
        not np.issubdtype(y_arr.dtype, np.floating)):
        print("invalid input")
        return

    # Check same length
    if len(x_arr) != len(y_arr):
        print("arrays are not the same size.")
        return

    # Replace NaN values in x_arr
    masked_x = np.ma.masked_invalid(x_arr)
    mean_x = masked_x.mean()
    x_arr = masked_x.filled(mean_x)

    # Replace NaN values in y_arr
    masked_y = np.ma.masked_invalid(y_arr)
    mean_y = masked_y.mean()
    y_arr = masked_y.filled(mean_y)

    # Create scatter plot
    plt.scatter(x_arr, y_arr)
    plt.title("Scatter Plot")
    plt.xlabel("x values")
    plt.ylabel("y values")
    plt.show()
```

```python
import numpy as np
x = np.array([1.0, 2.0, 3.0, 4.0])
y = np.array([2.0, 4.0, 6.0, 8.0])
create_scatter_plot(x, y)
```

```python
import numpy as np

def calculate_regression(x, y):
    # Validate inputs
    if (not isinstance(x, np.ndarray) or
        not isinstance(y, np.ndarray) or
        x.ndim != 1 or
        y.ndim != 1 or
        not np.issubdtype(x.dtype, np.floating) or
        not np.issubdtype(y.dtype, np.floating)):
        print("invalid input")
        return None

    # Check same length
    if len(x) != len(y):
        print("x and y are not the same length")
        return None

    # Make copies so original arrays are not modified
    x_copy = x.copy()
    y_copy = y.copy()

    # Impute missing values in x
    if np.isnan(x_copy).any():
        x_mean = np.nanmean(x_copy)
        x_copy[np.isnan(x_copy)] = x_mean

    # Impute missing values in y
    if np.isnan(y_copy).any():
        y_mean = np.nanmean(y_copy)
        y_copy[np.isnan(y_copy)] = y_mean

    # Calculate slope and intercept
    denominator = np.sum((x_copy - np.mean(x_copy)) ** 2)
    if denominator == 0:
        # Handle vertical line case (x values are all the same)
        return (np.nan, np.nan)

    slope = np.sum((x_copy - np.mean(x_copy)) * (y_copy - np.mean(y_copy))) / denominator

    intercept = np.mean(y_copy) - slope * np.mean(x_copy)

    return (slope, intercept)
```

```python
import numpy as np
x = np.array([1.0, 2.0, 3.0, 4.0])
y = np.array([2.0, 4.0, 5.0, 8.0])
calculate_regression(x, y)
```

```python
def plot_regression(x, y):
    # Call previous function for validation and regression calculation
    result = calculate_regression(x, y)

    # If input was invalid
    if result is None:
        return None

    slope, intercept = result

    # Check for vertical line
    if np.isnan(slope):
        print("vertical line detected")
        return None

    # Make copies so original arrays are not modified
    x_plot = x.copy()
    y_plot = y.copy()

    # Impute missing values
    if np.isnan(x_plot).any():
        x_plot[np.isnan(x_plot)] = np.nanmean(x_plot)

    if np.isnan(y_plot).any():
        y_plot[np.isnan(y_plot)] = np.nanmean(y_plot)

    # Create plot
    fig, ax = plt.subplots()

    # Scatter plot
    ax.scatter(x_plot, y_plot)

    # Regression line
    ax.axline(xy1=(0, intercept), slope=slope)

    # Labels and title
    ax.set_title("Linear Regression Plot")
    ax.set_xlabel("x")
    ax.set_ylabel("y")

    plt.show()
```

```python
import numpy as np
x = np.array([1.0, 2.0, 3.0, 4.0])
y = np.array([2.0, 4.0, 5.0, 8.0])
plot_regression(x, y)
```
